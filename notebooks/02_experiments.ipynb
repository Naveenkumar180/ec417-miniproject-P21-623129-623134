```python
import os; os.chdir('..') if os.path.basename(os.getcwd()) == 'notebooks' else None
import subprocess, sys
CONFIGS = [['--model','majority'], ['--model','rf'], ['--model','lstm'], ['--model','bilstm'],
           ['--model','cnnlstm'], ['--model','bilstm','--weighted'], ['--model','cnnlstm','--weighted']]
SEEDS = [0, 1, 2]
```

```python
for seed in SEEDS:
    for cfg in CONFIGS:
        subprocess.run([sys.executable, '-m', 'src.train', *cfg, '--seed', str(seed)], check=True)
```
