## Class balance
Why accuracy is misleading: what accuracy does the majority class get on test?

```python
y = S['y_test']; print('majority-class test accuracy:', (y == np.bincount(S['y_train']).argmax()).mean())
summary(S).set_index('split').div(summary(S).set_index('split')['n'], axis=0).round(4)
```

## Average waveform per class (±1 std)

```python
fig, ax = plt.subplots(1, 5, figsize=(16, 2.6), sharey=True)
for c in range(5):
    X = S['X_train'][S['y_train'] == c]
    m, s = X.mean(0), X.std(0)
    ax[c].plot(m); ax[c].fill_between(range(N_SAMPLES), m - s, m + s, alpha=.3); ax[c].set_title(f'{CLASS_NAMES[c]} (n={len(X)})')
plt.tight_layout(); plt.savefig('results/figures/class_mean_waveforms.png', dpi=150)
```

## Leakage checks
Exact duplicate beats between train and test, and NaNs.

```python
print('test beats identical to a training beat:', duplicate_report(S['X_train'], S['X_test']), 'of', len(S['X_test']))
print('NaNs:', np.isnan(S['X_train']).sum(), np.isnan(S['X_test']).sum())
print('value range:', S['X_train'].min(), S['X_train'].max())
```
